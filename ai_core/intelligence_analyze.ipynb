```python
import sqlite3
import pandas as pd
 
conn=sqlite3.connect("../trends.db")
df=pd.read_sql_query("SELECT * FROM videos", conn)
conn.close()
print(df.head())
print(df.info())
print(df.isna().sum())
```

```python

df['publish_date'] = pd.to_datetime(df['publish_date'], utc=True)


today = pd.Timestamp.now(tz='UTC')


df['video_age_day'] = (today - df['publish_date']).dt.days


print(df[['title', 'publish_date', 'video_age_day']].head())
```

```python

df.drop('publish_date', axis=1, inplace=True)


print(df.head())
```

```python

df['title_length'] = df['title'].str.len()


df['label_count'] = df['labels'].str.split(',').str.len()


print("=== NEW NUMERICAL FEATURES DERIVED FROM TEXT ===")
print(df[['title', 'title_length', 'label_count']].head())
```
